```python
import pandas as pd
import requests

url = 'https://api.worldbank.org/v2/country/NGA/indicator/FP.CPI.TOTL.ZG?format=json&per_page=100'
answer = requests.get(url).json()
data = pd.DataFrame(answer[1])
print(data[['date', 'value']].head(10))
```

```python
import matplotlib.pyplot as plt
data = data.iloc[::-1]
plt.plot(data.date.astype(int), data.value)
plt.show()
```

```python
import matplotlib.pyplot as plt
data = data.iloc[::-1]
plt.plot(data.date.astype(int), data.value)
plt.title('Nigeria inflation rate, 1960 to 2025')
plt.xlabel('Year')
plt.ylabel('Inflation, percent per year')
plt.show()
```

Nigeria inflation, 1960 to 2025. Data source: World Bank. This chart shows how fast prices rose each year in Nigeria. Inflation peaked above 70 percent in the mid-1990s, then settled around 10 to 15 percent for much of the 2000s and 2010s. It has climbed sharply since 2016, reaching about 33 percent in 2024 before easing to about 23 percent in 2025. Built with Python (pandas and matplotlib), pulling live data from the World Bank API.
